```python
%sql
USE CATALOG samplesuperstore;

CREATE OR REPLACE FUNCTION silverdata.run_data_eda(dataset STRING)
RETURNS STRING
LANGUAGE PYTHON
AS $$
def run_data_eda(dataset: str) -> str:
    from pyspark.sql import SparkSession
    spark = SparkSession.builder.getOrCreate()
    
    # Read from silver table instead of CSV
    df = spark.table(dataset)
    
    negative_orders = df.filter(df['Profit'] < 0).count()
    return f"Orders with negative profit: {negative_orders}"
$$;
```

```python
%sql
-- Step 1: Switch to the correct catalog
USE CATALOG samplesuperstore;

-- Step 2: Call the UC function with silverdata.orders table
SELECT silverdata.run_data_eda('silverdata.orders');

-- Step 3: Verify function registration (optional check)
SHOW FUNCTIONS IN silverdata;
```
